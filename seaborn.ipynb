# BAR PLOT

```python
# 3. Bar Plot

import seaborn as sns
import matplotlib.pyplot as plt

categories = ["A", "B", "C", "D"]
values = [20, 35, 30, 40]

sns.barplot(x=categories, y=values)

plt.show()
```

#  Heatmap


```python

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = {
    "Math": [80, 90, 70, 85],
    "Science": [85, 95, 75, 80],
    "English": [70, 80, 90, 75]
}

df = pd.DataFrame(data)

correlation = df.corr()

sns.heatmap(correlation, annot=True)

plt.show()
```
